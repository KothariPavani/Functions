```python
#1.What is Hypothesis Testing in Statistics?
Ans:
Hypothesis testing is a statistical method used to evaluate assumptions about a population using sample data. It involves testing a null hypothesis (H₀) against an alternative
hypothesis (H₁) by analyzing the likelihood of observed data. It helps in making data-driven decisions by determining statistical significance.
```

```python
#2.What is the Null Hypothesis, and how Does it differ from the Alternative Hypothesis?
Ans:
The null hypothesis (H₀) assumes no effect or difference, while the alternative hypothesis (H₁) suggests a significant effect exists. Hypothesis testing evaluates sample data to
decide whether to reject H₀ in favor of H₁ based on statistical evidence.
```

```python
#3.What is the Significance Level in Hypothesis Testing, and Why Is It Important?
Ans:
The **significance level (α)** in hypothesis testing is the probability threshold used to decide whether to reject the null hypothesis (H₀), commonly set at **0.05**. It represents
the risk of making a **Type 1 error** (rejecting a true H₀). A lower α reduces the likelihood of Type 1 errors but increases the chance of a **Type 2 error** (failing to reject a
false H₀). The chosen significance level directly impacts the sensitivity and reliability of hypothesis testing outcomes.
```

```python
#4.What Does a P-Value Represent in Hypothesis Testing?
Ans:
A **P-value** in hypothesis testing represents the probability of obtaining results as extreme as the observed data, assuming the null hypothesis (**H₀**) is true. It helps
determine the strength of evidence against H₀. A **lower P-value** indicates stronger evidence against H₀. If the P-value is **less than the significance level (α)**, H₀ is rejected,
suggesting the results are statistically significant.
```

```python
#5.How do you interpret the P-value in hypothesis testing?
Ans:
In hypothesis testing, a **P-value** ≤ **α** (e.g., 0.05) indicates significant results, leading to the rejection of the null hypothesis (**H₀**). This suggests strong evidence
against H₀. If the **P-value** > **α**, H₀ is not rejected, meaning there isn't enough evidence to conclude a significant effect or relationship in the data.
```

```python
#6.What are Type 1 and Type 2 errors in hypothesis testing?
Ans:
In hypothesis testing, a **Type 1 error** occurs when a true null hypothesis (**H₀**) is wrongly rejected, leading to a false positive. A **Type 2 error** happens when a false H₀
is not rejected, resulting in a false negative. Type 1 errors are controlled by the significance level (**α**), while Type 2 errors are reduced by increasing sample size.
```

```python
#7.What is the difference between a one-tailed and a two-tailed test in hypothesis testing?
Ans:
A **one-tailed test** evaluates if a parameter is significantly greater or smaller than a specific value, testing in one direction. It is used when a clear directional effect is
expected. A **two-tailed test** checks for deviations in both directions, assessing whether a parameter is significantly different from a value, without specifying the direction of
the effect.
```

```python
#8.What is the Z-test, and when is it used in hypothesis testing?
Ans:
A **Z-test** is a statistical test used to determine if there is a significant difference between sample and population means. It is applied when the sample size is large
(**n > 30**) and the population standard deviation is known. The Z-test assumes data follows a normal distribution and is commonly used for comparing means and proportions.
```

```python
#9.How do you calculate the Z-score, and what does it represent in hypothesis testing?
Ans:
The **Z-score** measures how many standard deviations a data point is from the population mean. It is calculated using the formula:
Z = (X - μ) / σ

Where:
- **X** = Sample value
- **μ** = Population mean
- **σ** = Population standard deviation

In hypothesis testing, the Z-score helps determine how unusual or extreme a data point is under the null hypothesis.
```

```python
#10. What is the T-distribution, and when should it be used instead of the normal distribution?
Ans:
The **T-distribution** is a probability distribution used in hypothesis testing when the sample size is small (**n < 30**) and the population standard deviation is unknown. It is
similar to the normal distribution but has heavier tails, reflecting greater variability. The T-distribution is used in **T-tests** for estimating population means and constructing
confidence intervals.
```

```python
#11.What is the difference between a Z-test and a T-test?
Ans:
A **Z-test** is used for large samples (**n > 30**) when the population standard deviation is known, assuming a normal distribution. In contrast, a **T-test** is applied for small
samples (**n < 30**) or when the population standard deviation is unknown. The T-test uses the **T-distribution** to account for increased variability in smaller datasets.
```

```python
#12.What is the T-test, and how is it used in hypothesis testing?
Ans:
A **T-test** is a statistical test used to compare the means of two groups to determine if there is a significant difference between them. It is applied when the sample size is
small (**n < 30**) and the population standard deviation is unknown. T-tests are commonly used in hypothesis testing for comparing sample means in independent, paired, or
one-sample scenarios.
```

```python
#13.What is the relationship between Z-test and T-test in hypothesis testing?
Ans:
In hypothesis testing, both **Z-tests** and **T-tests** are used to compare sample means to a population mean or between groups. A **Z-test** is used for large samples (**n > 30**)
with a known population standard deviation, assuming a normal distribution. A **T-test** is applied for small samples (**n < 30**) or when the population standard deviation is
unknown, using the T-distribution.
```

```python
#14.What is a confidence interval, and how is it used to interpret statistical results?
Ans:
A **confidence interval (CI)** is a range of values used to estimate the true population parameter with a specified level of confidence, often **95%** or **99%**. It provides a
likely range for the population mean based on sample data. A wider CI indicates more uncertainty, while a narrower CI suggests more precise estimates, aiding in result interpretation.
```

```python
#15.What is the margin of error, and how does it affect the confidence interval?
Ans:
The **margin of error (MOE)** is the maximum expected difference between the sample estimate and the true population parameter. It is calculated using the standard error and a
critical value (Z or T-score). A larger margin of error results in a wider **confidence interval**, indicating less precision, while a smaller MOE provides a narrower, more
precise interval.
```

```python
#16.How is Bayes' Theorem used in statistics, and what is its significance?
Ans:
**Bayes' Theorem** is used in statistics to calculate conditional probabilities by updating prior knowledge with new evidence. It is significant in decision-making, especially in
fields like medical diagnosis, spam filtering, and fraud detection. The formula is:
P(A|B) = (P(B|A) * P(A)) / P(B)

Where \(P(A|B)\) is the updated probability of event A given B has occurred.
```

```python
#17.What is the Chi-square distribution, and when is it used?
Ans:
The **Chi-square distribution** is a probability distribution used for categorical data analysis. It is applied in hypothesis testing to assess the relationship between variables.
Common applications include the **Chi-square test for independence** and the **goodness-of-fit test**. It is used when data are in frequency or count form and helps determine
if observed data differ significantly from expected data.
```

```python
#18.What is the Chi-square goodness of fit test, and how is it applied?
Ans:
The **Chi-square goodness of fit test** evaluates whether observed categorical data align with expected distributions. It is applied by comparing the observed frequencies with the
expected frequencies under a specified hypothesis. The test statistic is calculated using:
Chi-square = Σ((O - E)^2) / E

Where O_i = Observed frequency, and E_i = Expected frequency. A significant result suggests a poor fit to the expected distribution.
```

```python
#19.What is the F-distribution, and when is it used in hypothesis testing?
Ans:
The **F-distribution** is a probability distribution used primarily in hypothesis testing to compare variances between groups. It is applied in tests like **ANOVA (Analysis of
Variance)** and the **F-test** to determine if the variances of two or more datasets are significantly different. The F-distribution is asymmetric and right-skewed, with values
always being non-negative.
```

```python
#20.What is an ANOVA test, and what are its assumptions?
Ans:
An **ANOVA (Analysis of Variance)** test compares the means of three or more groups to determine if there are significant differences between them. It is commonly used in
experimental research. The assumptions of ANOVA include:
1. **Normality**: Data in each group should be approximately normally distributed.
2. **Homogeneity of Variances**: Variances across groups should be equal.
3. **Independence**: Observations should be independent of each other.
```

```python
#21.What are the different types of ANOVA tests?
Ans:
The main types of **ANOVA tests** include:

1. **One-Way ANOVA**: Compares the means of three or more independent groups based on one factor.
2. **Two-Way ANOVA**: Examines the effect of two independent factors on a dependent variable.
3. **Repeated Measures ANOVA**: Analyzes data where the same subjects are tested under different conditions or over time.
```

```python
#22.What is the F-test, and how does it relate to hypothesis testing?
Ans:
An **F-test** is a statistical test used to compare the variances of two or more groups to assess if they are significantly different. It is commonly applied in **ANOVA** and
**regression analysis**. In hypothesis testing, the F-test evaluates the null hypothesis that group variances are equal. A high F-value may lead to rejecting the null hypothesis,
indicating significant differences.
```

```python
#1.Write a Python program to perform a Z-test for comparing a sample mean to a known population mean and interpret the results.
import numpy as np
import scipy.stats as stats

def z_test(sample_data, population_mean, population_std):
    sample_mean = np.mean(sample_data)
    sample_size = len(sample_data)

    # Calculate Z-score
    z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

    # Calculate P-value (two-tailed)
    p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

    print(f"Sample Mean: {sample_mean}")
    print(f"Z-Score: {z_score}")
    print(f"P-Value: {p_value}")

    # Interpretation
    alpha = 0.05
    if p_value < alpha:
        print("Reject the null hypothesis: There is a significant difference.")
    else:
        print("Fail to reject the null hypothesis: No significant difference.")

# Example data
sample_data = [55, 60, 58, 62, 59, 57, 63, 61]
population_mean = 60
population_std = 4

z_test(sample_data, population_mean, population_std)
```

```python
#4.Perform a two-tailed Z-test using Python and visualize the decision region on a plot.
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

def two_tailed_z_test(sample_data, population_mean, population_std):
    sample_size = len(sample_data)
    sample_mean = np.mean(sample_data)
    sample_std = np.std(sample_data, ddof=1)

    # Calculate Z-score
    z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

    # Calculate P-value for two-tailed test
    p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

    print(f"Sample Mean: {sample_mean:.2f}")
    print(f"Z-score: {z_score:.2f}")
    print(f"P-value: {p_value:.4f}")

    # Decision
    alpha = 0.05
    if p_value < alpha:
        print("Reject the null hypothesis: Significant difference detected.")
    else:
        print("Fail to reject the null hypothesis: No significant difference detected.")

    # Visualization of decision regions
    x = np.linspace(-4, 4, 1000)
    y = stats.norm.pdf(x, 0, 1)

    # Plot the normal distribution
    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label='Standard Normal Distribution')

    # Shade the rejection regions
    critical_value = stats.norm.ppf(1 - alpha / 2)
    plt.fill_between(x, y, where=(x <= -critical_value) | (x >= critical_value), color='red', alpha=0.5, label='Rejection Region')

    # Plot the Z-score
    plt.axvline(z_score, color='blue', linestyle='--', label=f'Z-Score = {z_score:.2f}')

    # Labels and legend
    plt.title('Two-Tailed Z-Test with Decision Region')
    plt.xlabel('Z-Score')
    plt.ylabel('Probability Density')
    plt.legend()
    plt.grid(True)
    plt.show()

# Example: Generate random sample data
np.random.seed(42)
sample_data = np.random.normal(loc=100, scale=15, size=30)
population_mean = 100
population_std = 15

# Perform two-tailed Z-test
two_tailed_z_test(sample_data, population_mean, population_std)
```

```python
#5.Create a Python function that calculates and visualizes Type 1 and Type 2 errors during hypothesis testing
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

def visualize_type1_type2_error(population_mean, sample_mean, population_std, sample_std, alpha=0.05, sample_size=30):
    # Calculate Z-critical value for Type 1 error (two-tailed)
    z_critical = stats.norm.ppf(1 - alpha / 2)

    # Calculate Type 2 error region using actual sample mean
    z_beta = (population_mean - sample_mean) / (sample_std / np.sqrt(sample_size))
    beta = stats.norm.cdf(z_critical - z_beta)

    # Plot standard normal distribution
    x = np.linspace(-4, 4, 1000)
    y = stats.norm.pdf(x, 0, 1)

    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label='Standard Normal Distribution', color='blue')

    # Type 1 Error (Alpha): Rejection region when H0 is true
    plt.fill_between(x, y, where=(x <= -z_critical) | (x >= z_critical), color='red', alpha=0.5, label='Type 1 Error (α)')

    # Type 2 Error (Beta): Region where we fail to reject H0 when H1 is true
    shifted_x = (x * (sample_std / population_std)) + (sample_mean - population_mean) / population_std
    plt.fill_between(shifted_x, y, where=(shifted_x > -z_critical) & (shifted_x < z_critical), color='orange', alpha=0.5, label='Type 2 Error (β)')

    # Plot decision boundary
    plt.axvline(-z_critical, color='black', linestyle='--', label=f'Z = {-z_critical:.2f}')
    plt.axvline(z_critical, color='black', linestyle='--', label=f'Z = {z_critical:.2f}')

    # Labels and legend
    plt.title('Visualization of Type 1 and Type 2 Errors')
    plt.xlabel('Z-Score')
    plt.ylabel('Probability Density')
    plt.legend()
    plt.grid(True)
    plt.show()

    print(f"Type 2 Error (β): {beta:.4f}")

# Example: Visualizing Errors
population_mean = 100
sample_mean = 102
population_std = 15
sample_std = 15
alpha = 0.05
sample_size = 30

visualize_type1_type2_error(population_mean, sample_mean, population_std, sample_std, alpha, sample_size)
```

```python
#9.Write a Python function to calculate the confidence interval for a sample mean and explain its significance.
import numpy as np
from scipy import stats

def calculate_confidence_interval(data, confidence=0.95):
    # Calculate sample mean and standard error
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)
    n = len(data)
    standard_error = sample_std / np.sqrt(n)

    # Find the critical value using the T-distribution for small samples
    t_critical = stats.t.ppf((1 + confidence) / 2, df=n-1)

    # Calculate margin of error
    margin_of_error = t_critical * standard_error

    # Calculate confidence interval
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    print(f"Sample Mean: {sample_mean:.2f}")
    print(f"Confidence Interval: ({lower_bound:.2f}, {upper_bound:.2f}) at {confidence*100:.1f}% confidence level")
    return lower_bound, upper_bound

# Example: Generate random sample data
np.random.seed(42)
sample_data = np.random.normal(loc=100, scale=15, size=30)

# Calculate Confidence Interval
calculate_confidence_interval(sample_data)
```

```python
#10.Write a Python program to calculate the margin of error for a given confidence level using sample data.
import numpy as np
from scipy import stats

def calculate_margin_of_error(data, confidence=0.95):
    # Calculate sample mean and standard deviation
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)
    n = len(data)
    standard_error = sample_std / np.sqrt(n)

    # Determine the critical value using the T-distribution
    t_critical = stats.t.ppf((1 + confidence) / 2, df=n-1)

    # Calculate Margin of Error
    margin_of_error = t_critical * standard_error

    print(f"Sample Mean: {sample_mean:.2f}")
    print(f"Sample Standard Deviation: {sample_std:.2f}")
    print(f"Margin of Error: {margin_of_error:.2f} at {confidence*100:.1f}% Confidence Level")
    return margin_of_error

# Example: Generate random sample data
np.random.seed(42)
sample_data = np.random.normal(loc=100, scale=15, size=30)

# Calculate Margin of Error
calculate_margin_of_error(sample_data)
```

```python
#11.Implement a Bayesian inference method using Bayes' Theorem in Python and explain the process.
def bayes_theorem(prior_A, likelihood_B_given_A, likelihood_B_given_not_A):
    # Calculate the probability of B (evidence)
    evidence_B = (likelihood_B_given_A * prior_A) + (likelihood_B_given_not_A * (1 - prior_A))

    # Calculate Posterior Probability using Bayes' Theorem
    posterior_A_given_B = (likelihood_B_given_A * prior_A) / evidence_B

    print(f"Prior Probability (P(A)): {prior_A:.4f}")
    print(f"Likelihood (P(B|A)): {likelihood_B_given_A:.4f}")
    print(f"Likelihood (P(B|¬A)): {likelihood_B_given_not_A:.4f}")
    print(f"Evidence (P(B)): {evidence_B:.4f}")
    print(f"Posterior Probability (P(A|B)): {posterior_A_given_B:.4f}")
    return posterior_A_given_B

# Example: Medical Test Scenario
prior_A = 0.01  # Probability that a person has a disease
likelihood_B_given_A = 0.95  # Probability of a positive test result if the person has the disease
likelihood_B_given_not_A = 0.05  # Probability of a positive test result if the person doesn't have the disease

# Calculate and print the result
bayes_theorem(prior_A, likelihood_B_given_A, likelihood_B_given_not_A)
```

```python
#12.Perform a Chi-square test for independence between two categorical variables in Python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

def chi_square_test(data):
    # Create a contingency table
    contingency_table = pd.crosstab(data['Variable1'], data['Variable2'])

    # Perform Chi-Square Test
    chi2, p_value, dof, expected = chi2_contingency(contingency_table)

    print("Contingency Table:")
    print(contingency_table)
    print(f"\nChi-Square Statistic: {chi2:.2f}")
    print(f"Degrees of Freedom: {dof}")
    print(f"P-Value: {p_value:.4f}")

    # Interpretation
    alpha = 0.05
    if p_value < alpha:
        print("Reject the null hypothesis: Variables are dependent.")
    else:
        print("Fail to reject the null hypothesis: Variables are independent.")

# Example Data (Simulated)
data = pd.DataFrame({
    'Variable1': np.random.choice(['Male', 'Female'], size=100),
    'Variable2': np.random.choice(['Yes', 'No'], size=100)
})

# Perform Chi-Square Test
chi_square_test(data)
```

```python
#14.Perform a goodness-of-fit test using Python to compare the observed data to an expected distribution.
import numpy as np
from scipy.stats import chisquare

def goodness_of_fit_test(observed_data, expected_data):
    # Perform Chi-Square Goodness-of-Fit Test
    chi2_stat, p_value = chisquare(f_obs=observed_data, f_exp=expected_data)

    print(f"Observed Data: {observed_data}")
    print(f"Expected Data: {expected_data}")
    print(f"Chi-Square Statistic: {chi2_stat:.2f}")
    print(f"P-Value: {p_value:.4f}")

    # Interpretation
    alpha = 0.05
    if p_value < alpha:
        print("Reject the null hypothesis: The observed data does not fit the expected distribution.")
    else:
        print("Fail to reject the null hypothesis: The observed data fits the expected distribution.")

# Example Data
observed_data = np.array([50, 30, 20])
expected_data = np.array([40, 40, 20])

# Perform Goodness-of-Fit Test
goodness_of_fit_test(observed_data, expected_data)
```

```python
#16. Implement an F-test using Python to compare the variances of two random samples.
import numpy as np
from scipy.stats import f

def perform_f_test(sample1, sample2):
    # Calculate sample variances
    var1 = np.var(sample1, ddof=1)
    var2 = np.var(sample2, ddof=1)

    # Calculate the F-statistic (larger variance / smaller variance)
    if var1 > var2:
        f_stat = var1 / var2
        dfn = len(sample1) - 1
        dfd = len(sample2) - 1
    else:
        f_stat = var2 / var1
        dfn = len(sample2) - 1
        dfd = len(sample1) - 1

    # Calculate the P-value
    p_value = 1 - f.cdf(f_stat, dfn, dfd)

    print(f"Variance of Sample 1: {var1:.2f}")
    print(f"Variance of Sample 2: {var2:.2f}")
    print(f"F-Statistic: {f_stat:.2f}")
    print(f"P-Value: {p_value:.4f}")

    # Interpret the result
    alpha = 0.05
    if p_value < alpha:
        print("Reject the null hypothesis: Variances are significantly different.")
    else:
        print("Fail to reject the null hypothesis: Variances are not significantly different.")

# Generate random sample data
np.random.seed(42)
sample1 = np.random.normal(loc=50, scale=10, size=30)
sample2 = np.random.normal(loc=50, scale=15, size=30)

# Perform the F-test
perform_f_test(sample1, sample2)
```

```python
#17.Write a Python program to perform an ANOVA test to compare means between multiple groups and interpret the results.
import numpy as np
from scipy import stats

def perform_anova_test(*groups):
    # Perform One-Way ANOVA
    f_stat, p_value = stats.f_oneway(*groups)

    print(f"F-Statistic: {f_stat:.2f}")
    print(f"P-Value: {p_value:.4f}")

    # Interpret Results
    alpha = 0.05
    if p_value < alpha:
        print("Reject the null hypothesis: At least one group mean is significantly different.")
    else:
        print("Fail to reject the null hypothesis: No significant difference between group means.")

# Generate random sample data for three groups
np.random.seed(42)
group1 = np.random.normal(loc=50, scale=10, size=30)
group2 = np.random.normal(loc=55, scale=10, size=30)
group3 = np.random.normal(loc=60, scale=10, size=30)

# Perform ANOVA Test
perform_anova_test(group1, group2, group3)
```

```python
#18.Perform a one-way ANOVA test using Python to compare the means of different groups and plot the results.
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

def perform_anova_test_and_plot(group_data):
    # Perform One-Way ANOVA
    f_stat, p_value = stats.f_oneway(*group_data.values())
    print(f"F-Statistic: {f_stat:.2f}")
    print(f"P-Value: {p_value:.4f}")

    # Interpret Results
    alpha = 0.05
    if p_value < alpha:
        print("Reject the null hypothesis: At least one group mean is significantly different.")
    else:
        print("Fail to reject the null hypothesis: No significant difference between group means.")

    # Convert data for visualization
    data = []
    for group_name, values in group_data.items():
        data.extend([(group_name, val) for val in values])
    df = pd.DataFrame(data, columns=["Group", "Value"])

    # Plot the data using boxplots
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Group', y='Value', data=df, palette='Set2')
    plt.title('One-Way ANOVA: Group Comparison')
    plt.xlabel('Groups')
    plt.ylabel('Values')
    plt.grid(True)
    plt.show()

# Generate random sample data for three groups
np.random.seed(42)
group_data = {
    'Group 1': np.random.normal(loc=50, scale=10, size=30),
    'Group 2': np.random.normal(loc=55, scale=10, size=30),
    'Group 3': np.random.normal(loc=60, scale=10, size=30)
}

# Perform ANOVA and Plot Results
perform_anova_test_and_plot(group_data)
```

```python
#19.Write a Python function to check the assumptions (normality, independence, and equal variance) for ANOVA.
import numpy as np
import pandas as pd
from scipy import stats
import seaborn as sns
import matplotlib.pyplot as plt

def check_anova_assumptions(group_data):
    print("Checking Assumptions for ANOVA...\n")

    # Convert data to DataFrame for easier handling
    data = []
    for group_name, values in group_data.items():
        data.extend([(group_name, val) for val in values])
    df = pd.DataFrame(data, columns=["Group", "Value"])

    # 1. Normality Test using Shapiro-Wilk
    print("1. Normality Test (Shapiro-Wilk Test)")
    for group_name, values in group_data.items():
        stat, p_value = stats.shapiro(values)
        print(f"{group_name} -> W-Statistic: {stat:.2f}, P-Value: {p_value:.4f}")
        if p_value < 0.05:
            print(f"  - {group_name} is not normally distributed.")
        else:
            print(f"  - {group_name} is normally distributed.")

    # 2. Independence (Assumed through experimental design)
    print("\n2. Independence Assumption")
    print("  - Ensure data collection was independent (This is typically assessed from experimental design).")

    # 3. Homogeneity of Variance using Levene's Test
    print("\n3. Equal Variance Test (Levene's Test)")
    stat, p_value = stats.levene(*group_data.values())
    print(f"Levene's Statistic: {stat:.2f}, P-Value: {p_value:.4f}")
    if p_value < 0.05:
        print("  - Variances are significantly different. The assumption of equal variance is violated.")
    else:
        print("  - Variances are approximately equal.")

    # Visualization using Boxplot
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Group', y='Value', data=df, palette='Set2')
    plt.title('Boxplot for Group Comparison')
    plt.grid(True)
    plt.show()

# Generate Random Data for Three Groups
np.random.seed(42)
group_data = {
    'Group 1': np.random.normal(loc=50, scale=10, size=30),
    'Group 2': np.random.normal(loc=55, scale=10, size=30),
    'Group 3': np.random.normal(loc=60, scale=10, size=30)
}

# Perform Assumption Checks
check_anova_assumptions(group_data)
```

```python
#24.Perform a hypothesis test for population variance using a Chi-square distribution and interpret the results.
import numpy as np
from scipy.stats import chi2

def perform_chi_square_variance_test(data, population_variance):
    # Step 1: Calculate sample variance
    sample_size = len(data)
    sample_variance = np.var(data, ddof=1)

    # Step 2: Calculate the Chi-square test statistic
    chi_square_stat = (sample_size - 1) * sample_variance / population_variance

    # Step 3: Calculate the p-value (two-tailed test)
    p_value = 2 * min(chi2.cdf(chi_square_stat, df=sample_size-1),
                       1 - chi2.cdf(chi_square_stat, df=sample_size-1))

    # Step 4: Print results
    print(f"Sample Variance: {sample_variance:.2f}")
    print(f"Chi-Square Statistic: {chi_square_stat:.2f}")
    print(f"P-Value: {p_value:.4f}")

    # Step 5: Interpret results
    alpha = 0.05
    if p_value < alpha:
        print("Reject the null hypothesis: Significant difference in variances.")
    else:
        print("Fail to reject the null hypothesis: No significant difference in variances.")

# Generate random data (simulating a normal distribution)
np.random.seed(42)
sample_data = np.random.normal(loc=50, scale=10, size=30)
population_variance = 100  # Assume the population variance is 100

# Perform Chi-square variance test
perform_chi_square_variance_test(sample_data, population_variance)
```

```python
#25.Write a Python script to perform a Z-test for comparing proportions between two datasets or groups.
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

def perform_z_test_for_proportions(success_a, size_a, success_b, size_b):
    # Step 1: Calculate proportions
    prop_a = success_a / size_a
    prop_b = success_b / size_b

    # Step 2: Perform Z-test for proportions
    count = np.array([success_a, success_b])
    nobs = np.array([size_a, size_b])
    z_stat, p_value = proportions_ztest(count, nobs)

    # Step 3: Print results
    print(f"Proportion A: {prop_a:.4f}, Proportion B: {prop_b:.4f}")
    print(f"Z-Statistic: {z_stat:.4f}")
    print(f"P-Value: {p_value:.4f}")

    # Step 4: Interpret results
    alpha = 0.05
    if p_value < alpha:
        print("Reject the null hypothesis: Significant difference between proportions.")
    else:
        print("Fail to reject the null hypothesis: No significant difference between proportions.")

# Example Data
success_a = 45
size_a = 100
success_b = 30
size_b = 100

# Perform Z-test
perform_z_test_for_proportions(success_a, size_a, success_b, size_b)
```

```python
#26.Implement an F-test for comparing the variances of two datasets, then interpret and visualize the results.
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f

def perform_f_test(data1, data2):
    # Calculate variances
    var1 = np.var(data1, ddof=1)
    var2 = np.var(data2, ddof=1)

    # Perform F-test (larger variance / smaller variance)
    if var1 > var2:
        f_statistic = var1 / var2
        df1, df2 = len(data1) - 1, len(data2) - 1
    else:
        f_statistic = var2 / var1
        df1, df2 = len(data2) - 1, len(data1) - 1

    # Calculate P-value (two-tailed test)
    p_value = 2 * (1 - f.cdf(f_statistic, df1, df2))

    # Print results
    print(f"Variance of Dataset 1: {var1:.2f}")
    print(f"Variance of Dataset 2: {var2:.2f}")
    print(f"F-Statistic: {f_statistic:.4f}")
    print(f"P-Value: {p_value:.4f}")

    if p_value < 0.05:
        print("Reject the null hypothesis: Significant difference in variances.")
    else:
        print("Fail to reject the null hypothesis: No significant difference in variances.")

    # Visualize the F-Distribution
    x = np.linspace(0, 5, 1000)
    y = f.pdf(x, df1, df2)

    plt.figure(figsize=(8, 5))
    plt.plot(x, y, label=f"F-Distribution (df1={df1}, df2={df2})", color='blue')
    plt.axvline(f_statistic, color='red', linestyle='--', label=f"F-Statistic = {f_statistic:.4f}")
    plt.fill_between(x, y, where=(x >= f_statistic), color='red', alpha=0.3, label='Rejection Region')
    plt.title('F-Test for Variance Comparison')
    plt.xlabel('F-Statistic')
    plt.ylabel('Probability Density')
    plt.legend()
    plt.grid(True)
    plt.show()

# Example: Generate Random Data
np.random.seed(42)
data1 = np.random.normal(loc=50, scale=10, size=30)
data2 = np.random.normal(loc=55, scale=15, size=30)

# Perform F-Test and Visualize
perform_f_test(data1, data2)
```
